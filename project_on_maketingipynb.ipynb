```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("/content/marketing.csv")
```

```python
print("First 5 rows of the DataFrame:")
print(df.head())

print("\nShape of the DataFrame:")
print(df.shape)

print("\nData types of each column:")
print(df.info())
```

```python
print("Missing values before imputation:")
print(df.isnull().sum())

# Impute missing values in 'MINIMUM_PAYMENTS' with its median
df['MINIMUM_PAYMENTS'].fillna(df['MINIMUM_PAYMENTS'].median(), inplace=True)

# Impute missing values in 'CREDIT_LIMIT' with its median
df['CREDIT_LIMIT'].fillna(df['CREDIT_LIMIT'].median(), inplace=True)

print("\nMissing values after imputation:")
print(df.isnull().sum())
```

**Reasoning**:
The previous code generated a `FutureWarning` regarding the use of `inplace=True` with `fillna`. To address this warning and ensure proper imputation, I will modify the `fillna` calls to directly assign the result back to the DataFrame columns instead of using `inplace=True`.



```python
print("Missing values before imputation:")
print(df.isnull().sum())

# Impute missing values in 'MINIMUM_PAYMENTS' with its median
df['MINIMUM_PAYMENTS'] = df['MINIMUM_PAYMENTS'].fillna(df['MINIMUM_PAYMENTS'].median())

# Impute missing values in 'CREDIT_LIMIT' with its median
df['CREDIT_LIMIT'] = df['CREDIT_LIMIT'].fillna(df['CREDIT_LIMIT'].median())

print("\nMissing values after imputation:")
print(df.isnull().sum())
```

**Reasoning**:
To understand the central tendency, dispersion, and shape of the data's distribution, I will generate descriptive statistics for all numerical columns in the DataFrame using the `.describe()` method.



```python
print("Descriptive statistics for numerical columns:")
print(df.describe())
```

**Reasoning**:
To begin the exploratory data analysis, I will generate histograms for five key numerical columns: 'BALANCE', 'PURCHASES', 'CREDIT_LIMIT', 'PAYMENTS', and 'CASH_ADVANCE'. This will help visualize the distribution of each variable and identify any skewness or unusual patterns.



```python
numerical_cols = ['BALANCE', 'PURCHASES', 'CREDIT_LIMIT', 'PAYMENTS', 'CASH_ADVANCE']

plt.figure(figsize=(15, 10))
for i, col in enumerate(numerical_cols):
    plt.subplot(2, 3, i + 1) # Arrange plots in 2 rows, 3 columns
    sns.histplot(df[col], kde=True)
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```

**Reasoning**:
Following the previous step of generating histograms, I will now create box plots for the same five numerical columns to visualize their spread and identify potential outliers. This fulfills the second instruction of the subtask.



```python
numerical_cols = ['BALANCE', 'PURCHASES', 'CREDIT_LIMIT', 'PAYMENTS', 'CASH_ADVANCE']

plt.figure(figsize=(15, 10))
for i, col in enumerate(numerical_cols):
    plt.subplot(2, 3, i + 1) # Arrange plots in 2 rows, 3 columns
    sns.boxplot(y=df[col])
    plt.title(f'Box Plot of {col}')
    plt.ylabel(col)
plt.tight_layout()
plt.show()
```

**Reasoning**:
As per the third instruction of the subtask, I will now generate scatter plots for at least three pairs of numerical columns to visualize the relationships between them. I will choose 'BALANCE' vs 'PURCHASES', 'CREDIT_LIMIT' vs 'BALANCE', and 'PAYMENTS' vs 'PURCHASES' to show potential correlations.

As per the fourth instruction of the subtask, I will now compute and visualize the correlation matrix for all numerical columns using a heatmap. This will help identify the strength and direction of linear relationships between variables.
